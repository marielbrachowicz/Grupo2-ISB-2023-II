Una FFT global transforma una señal completa al dominio de la frecuencia, permitiendo conocer qué frecuencias están presentes, pero pierde la información temporal. Esto significa que si aparece un evento en un momento específico de la señal, la FFT indicará que esa frecuencia existe, pero no podrá decir exactamente cuándo ocurrió.

En cambio, la STFT analiza la señal utilizando ventanas pequeñas de tiempo, realizando una FFT en cada segmento. De esta manera conserva información temporal y frecuencial, permitiendo identificar en qué momento aparecen ciertos eventos o cambios dentro de la señal.
