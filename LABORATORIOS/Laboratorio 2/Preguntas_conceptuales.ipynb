1. ¿Qué es PhysioNet?
PhysioNet es una plataforma que almacena y comparte bases de datos de señales biomédicas reales, como electrocardiogramas (ECG), señales cerebrales (EEG), respiración, entre otras. Permite que estudiantes e investigadores puedan acceder a datos para analizarlos y desarrollar métodos en medicina e ingeniería biomédica.

2. ¿Qué información proporciona record.p_signal?
Contiene la señal fisiológica registrada en formato numérico, es decir, los valores de amplitud de la señal para cada instante de tiempo. Estos datos representan la evolución de la señal biomédica después de haber sido convertida a unidades físicas.

3. ¿Qué representa la frecuencia de muestreo fs?
La frecuencia de muestreo (fs) indica cuántas muestras de una señal se toman por segundo. Se mide en Hertz (Hz). Por ejemplo: fs = 250 Hz significa que se registran 250 valores de la señal cada segundo. Mientras mayor sea fs, mayor detalle temporal tendrá la señal. 

4. ¿Qué representa la componente DC de una señal?
La compontente DC representa el valor promedio o nivel constante de una señal. Es la parte de la señal que no cambia con el tiempo y corresponde a la frecuencia 0Hz. Por ejemplo, si una señal de ECG está desplazada hacia arriba debido a un nivel base, eso deplazamiento corresponde a la componente DC. 

5. ¿Por qué se resta la media antes de analizar el escpectro AC?
Porque la media representa la componente DC de la señal y puede dominar el espectro de frecuencias. AL restar la media (componente AC), observado las frecuencias realmente relacionadas con los cambios fisiológicos.

6. ¿Qué información proporciona la FFT?
La FFT (Transformada Rápida de Fourier) permite conocer qué frecuencias están presentes en una señal y con qué intensidad. Convierte una señal del dominio del tiempo al dominio de la frecuencia. Por ejemplo, permite identificar ritmos cardíacos, frecuencias respiratorias o componentes periódicas dentro de una señal biomédica.

7. ¿Qué información adicional proporciona la STFT?
La STFT (Transformada de Fourier de Tiempo Corto) muestra cómo cambia el contenido frecuencial de una señal a lo largo del tiempo. A diferencia de la FFT, permite saber en qué momento aparecen ciertas frecuencias. Es útil para señales biomédicas que no son constantes, como ECG o EEG.

8. ¿Qué relación existe entre fs, número de muestras y duración?
La duración de una señal depende de la cantidad de muestras y la frecuencia de muestreo: Duración = número de muestras/fs. Si aumenta fs, se toman más muestras por segundo y la duración representada por un mismo número de muestras disminuye.

9. ¿Qué efecto tiene aumentar el tamaño de ventana de la STFT?
Al aumentar el tamaño de ventana: Mejora la resolución en frecuencia (se distinguen mejor frecuencias cercanas). Disminuye la resolución temporal (se pierde precisión sobre cuándo ocurre un evento). Es decir, se obtiene más detalle de frecuencias pero menos detalle de cambios rápidos.

10. ¿Por qué no es suficiente utilizar únicamente una representación temporal para analizar completamente una señal biomédica?
Porque una representación temporal solo muestra cómo cambia la amplitud con el tiempo, pero no permite identificar las frecuencias presentes en la señal. Muchas características importantes de señales biomédicas están relacionadas con sus componentes frecuenciales, por lo que es necesario complementar el análisis temporal con herramientas como FFT o STFT.
