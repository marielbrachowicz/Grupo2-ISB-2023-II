```python
import wfdb
import numpy as np
import matplotlib.pyplot as plt
from scipy import signal
```

```python
DATABASE = "nsrdb"
RECORD = "16420"
record = wfdb.rdrecord(RECORD, pn_dir = DATABASE)
print("Registro cargado")
```

```python
fs = record.fs
channels = record.sig_name
print("Frecuencia de muestreo:", fs)
print("Canales disponibles:", channels)
```

```python
CHANNEL = 0
x = record.p_signal[:, CHANNEL]
print("Número de muestras:", len(x))
```

```python
t = np.arange(len(x))/fs
plt.figure(f"ECG Registro {RECORD}")
plt.xlabel("Tiempo (s)")
plt.ylabel("Amplitud")
plt.grid(True)
plt.show()
```

```python
N = len(x)
X = np.fft.fft(x)
f = np.fft.fftfreq(N, 1/fs)
```

```python
dc = np.mean(x)
print("Componente DC:", dc)
x_dc = x - dc
x_dc = np.fft.fft(x_dc)

f = np.fft.fftfreq(len(x_dc), 1/fs)
```

```python
positive = f>= 0
plt.figure(figsize=(10, 4))
plt.plot(f[positive], np.abs(x_dc[positive]))
plt.xlim(0, 20)
plt.xlabel("Frecuencia (Hz)")
plt.ylabel("Magnitud")
plt.title("FFT sin componente DC")
plt.grid(True)
plt.show()
```

Conclusión
Se seleccionó el registro 16420 de la base PhysioNet y se analizó su señal ECG en los dominios temporal y frecuencial. Primero se identificó la frecuencia de muestreo y el canal utilizado. La representación temporal permitió observar la variación de la señal cardíaca a lo largo del tiempo.

Posteriormente se calculó la FFT para analizar las componentes frecuenciales de la señal. Finalmente, se eliminó la componente DC mediante la resta del valor promedio de la señal, permitiendo observar únicamente las variaciones dinámicas del ECG y mejorar la interpretación del espectro.
